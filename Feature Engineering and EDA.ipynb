```python
# ============================================================
# PROJECT 3: ADULT INCOME
# FEATURE ENGINEERING AND EXPLORATORY DATA ANALYSIS
# ============================================================

# ------------------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------------------

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option("display.max_columns", None)

# ------------------------------------------------------------
# 2. LOAD DATASET
# ------------------------------------------------------------

from google.colab import files

uploaded = files.upload()

# Change the filename if your uploaded file has a different name
df = pd.read_csv("adult.csv")

# ------------------------------------------------------------
# 3. BASIC DATASET INFORMATION
# ------------------------------------------------------------

print("First 5 rows:")
display(df.head())

print("\nLast 5 rows:")
display(df.tail())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

# ------------------------------------------------------------
# 4. STATISTICAL SUMMARY
# ------------------------------------------------------------

print("\nStatistical Summary:")
display(df.describe())

print("\nCategorical Summary:")
display(df.describe(include="object"))

# ------------------------------------------------------------
# 5. CLEAN COLUMN NAMES
# ------------------------------------------------------------

# Remove leading/trailing spaces from column names
df.columns = df.columns.str.strip()

print("\nCleaned Column Names:")
print(df.columns.tolist())

# ------------------------------------------------------------
# 6. CHECK MISSING VALUES
# ------------------------------------------------------------

print("\nMissing Values:")
print(df.isnull().sum())

print("\nMissing Value Percentage:")
missing_percentage = (df.isnull().sum() / len(df)) * 100
print(missing_percentage)

# ------------------------------------------------------------
# 7. HANDLE '?' VALUES
# ------------------------------------------------------------

# In the Adult dataset, missing categorical values
# are often represented using '?'

df = df.replace("?", np.nan)

print("\nMissing Values After Replacing '?' with NaN:")
print(df.isnull().sum())

# ------------------------------------------------------------
# 8. DUPLICATE CHECK
# ------------------------------------------------------------

print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())

# ------------------------------------------------------------
# 9. REMOVE DUPLICATES
# ------------------------------------------------------------

df = df.drop_duplicates()

print("\nShape After Removing Duplicates:")
print(df.shape)

# ------------------------------------------------------------
# 10. TARGET VARIABLE DISTRIBUTION
# ------------------------------------------------------------

# The target is usually named income
# Common values: <=50K and >50K

print("\nIncome Distribution:")
print(df["income"].value_counts())

plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="income")
plt.title("Income Distribution")
plt.xlabel("Income")
plt.ylabel("Number of People")
plt.show()

# ------------------------------------------------------------
# 11. INCOME PERCENTAGE
# ------------------------------------------------------------

income_percentage = df["income"].value_counts(normalize=True) * 100

print("\nIncome Percentage:")
print(income_percentage)

# ------------------------------------------------------------
# 12. AGE DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="age", bins=30, kde=True)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

# ------------------------------------------------------------
# 13. AGE VS INCOME
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="income", y="age")
plt.title("Age Distribution by Income")
plt.xlabel("Income")
plt.ylabel("Age")
plt.show()

# ------------------------------------------------------------
# 14. EDUCATION DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))
sns.countplot(
    data=df,
    y="education",
    order=df["education"].value_counts().index
)

plt.title("Education Distribution")
plt.xlabel("Count")
plt.ylabel("Education")
plt.show()

# ------------------------------------------------------------
# 15. EDUCATION VS INCOME
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))
sns.countplot(
    data=df,
    y="education",
    hue="income",
    order=df["education"].value_counts().index
)

plt.title("Education Level vs Income")
plt.xlabel("Count")
plt.ylabel("Education")
plt.show()

# ------------------------------------------------------------
# 16. HOURS PER WEEK DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))
sns.histplot(
    data=df,
    x="hours-per-week",
    bins=30,
    kde=True
)

plt.title("Hours Worked Per Week")
plt.xlabel("Hours Per Week")
plt.ylabel("Frequency")
plt.show()

# ------------------------------------------------------------
# 17. HOURS PER WEEK VS INCOME
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df,
    x="income",
    y="hours-per-week"
)

plt.title("Working Hours vs Income")
plt.xlabel("Income")
plt.ylabel("Hours Per Week")
plt.show()

# ------------------------------------------------------------
# 18. CAPITAL GAIN DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))
sns.histplot(
    data=df,
    x="capital-gain",
    bins=30,
    kde=True
)

plt.title("Capital Gain Distribution")
plt.xlabel("Capital Gain")
plt.ylabel("Frequency")
plt.show()

# ------------------------------------------------------------
# 19. CAPITAL LOSS DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))
sns.histplot(
    data=df,
    x="capital-loss",
    bins=30,
    kde=True
)

plt.title("Capital Loss Distribution")
plt.xlabel("Capital Loss")
plt.ylabel("Frequency")
plt.show()

# ------------------------------------------------------------
# 20. WORKCLASS DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))
sns.countplot(
    data=df,
    y="workclass",
    order=df["workclass"].value_counts().index
)

plt.title("Workclass Distribution")
plt.xlabel("Count")
plt.ylabel("Workclass")
plt.show()

# ------------------------------------------------------------
# 21. WORKCLASS VS INCOME
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))
sns.countplot(
    data=df,
    y="workclass",
    hue="income",
    order=df["workclass"].value_counts().index
)

plt.title("Workclass vs Income")
plt.xlabel("Count")
plt.ylabel("Workclass")
plt.show()

# ------------------------------------------------------------
# 22. MARITAL STATUS VS INCOME
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))
sns.countplot(
    data=df,
    y="marital-status",
    hue="income",
    order=df["marital-status"].value_counts().index
)

plt.title("Marital Status vs Income")
plt.xlabel("Count")
plt.ylabel("Marital Status")
plt.show()

# ------------------------------------------------------------
# 23. OCCUPATION VS INCOME
# ------------------------------------------------------------

plt.figure(figsize=(12, 7))
sns.countplot(
    data=df,
    y="occupation",
    hue="income",
    order=df["occupation"].value_counts().index
)

plt.title("Occupation vs Income")
plt.xlabel("Count")
plt.ylabel("Occupation")
plt.show()

# ------------------------------------------------------------
# 24. SEX VS INCOME
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))
sns.countplot(
    data=df,
    x="sex",
    hue="income"
)

plt.title("Gender vs Income")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()

# ------------------------------------------------------------
# 25. RACE VS INCOME
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))
sns.countplot(
    data=df,
    y="race",
    hue="income"
)

plt.title("Race vs Income")
plt.xlabel("Count")
plt.ylabel("Race")
plt.show()

# ------------------------------------------------------------
# 26. RELATIONSHIP VS INCOME
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))
sns.countplot(
    data=df,
    y="relationship",
    hue="income",
    order=df["relationship"].value_counts().index
)

plt.title("Relationship vs Income")
plt.xlabel("Count")
plt.ylabel("Relationship")
plt.show()

# ============================================================
# FEATURE ENGINEERING
# ============================================================

# ------------------------------------------------------------
# 27. CREATE AGE GROUP
# ------------------------------------------------------------

df["AgeGroup"] = pd.cut(
    df["age"],
    bins=[0, 18, 30, 45, 60, 100],
    labels=[
        "Teen",
        "Young Adult",
        "Adult",
        "Middle Age",
        "Senior"
    ]
)

print("\nAge Group:")
display(df[["age", "AgeGroup"]].head(10))

# ------------------------------------------------------------
# 28. CREATE CAPITAL TOTAL
# ------------------------------------------------------------

df["CapitalNet"] = df["capital-gain"] - df["capital-loss"]

print("\nCapitalNet:")
display(
    df[
        ["capital-gain", "capital-loss", "CapitalNet"]
    ].head()
)

# ------------------------------------------------------------
# 29. CREATE CAPITAL INVESTMENT FLAG
# ------------------------------------------------------------

df["HasCapitalGain"] = (
    df["capital-gain"] > 0
).astype(int)

print("\nHasCapitalGain:")
print(df["HasCapitalGain"].value_counts())

# ------------------------------------------------------------
# 30. CREATE WORKING HOURS CATEGORY
# ------------------------------------------------------------

df["WorkHoursCategory"] = pd.cut(
    df["hours-per-week"],
    bins=[0, 20, 40, 60, 100],
    labels=[
        "Part-time",
        "Full-time",
        "Overtime",
        "Extreme"
    ]
)

display(
    df[
        ["hours-per-week", "WorkHoursCategory"]
    ].head(10)
)

# ------------------------------------------------------------
# 31. CREATE EDUCATION LEVEL GROUP
# ------------------------------------------------------------

education_map = {
    "Preschool": 0,
    "1st-4th": 1,
    "5th-6th": 2,
    "7th-8th": 3,
    "9th": 4,
    "10th": 5,
    "11th": 6,
    "12th": 7,
    "HS-grad": 8,
    "Some-college": 9,
    "Assoc-voc": 10,
    "Assoc-acdm": 10,
    "Bachelors": 11,
    "Masters": 12,
    "Prof-school": 13,
    "Doctorate": 14
}

df["EducationLevel"] = df["education"].map(education_map)

print("\nEducation Level:")
display(
    df[
        ["education", "EducationLevel"]
    ].head(15)
)

# ------------------------------------------------------------
# 32. AGE GROUP VS INCOME
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))
sns.countplot(
    data=df,
    x="AgeGroup",
    hue="income"
)

plt.title("Age Group vs Income")
plt.xlabel("Age Group")
plt.ylabel("Count")
plt.show()

# ------------------------------------------------------------
# 33. WORK HOURS CATEGORY VS INCOME
# ------------------------------------------------------------

plt.figure(figsize=(9, 5))
sns.countplot(
    data=df,
    x="WorkHoursCategory",
    hue="income"
)

plt.title("Working Hours Category vs Income")
plt.xlabel("Working Hours Category")
plt.ylabel("Count")
plt.show()

# ------------------------------------------------------------
# 34. CAPITAL FLAG VS INCOME
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))
sns.countplot(
    data=df,
    x="HasCapitalGain",
    hue="income"
)

plt.title("Capital Gain Flag vs Income")
plt.xlabel("Has Capital Gain")
plt.ylabel("Count")
plt.show()

# ------------------------------------------------------------
# 35. EDUCATION LEVEL VS INCOME
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

sns.boxplot(
    data=df,
    x="income",
    y="EducationLevel"
)

plt.title("Education Level vs Income")
plt.xlabel("Income")
plt.ylabel("Education Level")
plt.show()

# ------------------------------------------------------------
# 36. CORRELATION MATRIX
# ------------------------------------------------------------

numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(12, 8))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")
plt.show()

# ------------------------------------------------------------
# 37. AVERAGE AGE BY INCOME
# ------------------------------------------------------------

print("\nAverage Age by Income:")

average_age_income = df.groupby("income")["age"].mean()

print(average_age_income)

# ------------------------------------------------------------
# 38. AVERAGE HOURS BY INCOME
# ------------------------------------------------------------

print("\nAverage Working Hours by Income:")

average_hours_income = df.groupby(
    "income"
)["hours-per-week"].mean()

print(average_hours_income)

# ------------------------------------------------------------
# 39. AVERAGE EDUCATION LEVEL BY INCOME
# ------------------------------------------------------------

print("\nAverage Education Level by Income:")

average_education_income = df.groupby(
    "income"
)["EducationLevel"].mean()

print(average_education_income)

# ------------------------------------------------------------
# 40. INCOME RATE BY EDUCATION
# ------------------------------------------------------------

education_income_rate = pd.crosstab(
    df["education"],
    df["income"],
    normalize="index"
) * 100

print("\nIncome Percentage by Education:")
display(education_income_rate)

# ------------------------------------------------------------
# 41. INCOME RATE BY SEX
# ------------------------------------------------------------

sex_income_rate = pd.crosstab(
    df["sex"],
    df["income"],
    normalize="index"
) * 100

print("\nIncome Percentage by Gender:")
display(sex_income_rate)

# ------------------------------------------------------------
# 42. FINAL DATASET INFORMATION
# ------------------------------------------------------------

print("\nFinal Dataset Shape:")
print(df.shape)

print("\nFinal Dataset Columns:")
print(df.columns.tolist())

print("\nFinal Missing Values:")
print(df.isnull().sum())

# ------------------------------------------------------------
# 43. SAVE FEATURE-ENGINEERED DATASET
# ------------------------------------------------------------

df.to_csv(
    "Adult_Income_Feature_Engineered.csv",
    index=False
)

print(
    "\nFeature-engineered dataset saved as "
    "Adult_Income_Feature_Engineered.csv"
)

# ------------------------------------------------------------
# 44. ML LIFECYCLE MAPPING
# ------------------------------------------------------------

print("""
============================================================
MACHINE LEARNING LIFECYCLE
============================================================

1. Problem Definition
   -> Understand and analyze factors associated with
      adult income.

2. Data Collection
   -> Adult Income dataset.

3. Data Understanding
   -> Examine rows, columns, data types and statistics.

4. Data Cleaning
   -> Replace '?' with missing values.
   -> Check duplicates and missing values.

5. Exploratory Data Analysis
   -> Analyze age, education, occupation, workclass,
      gender, marital status and working hours.

6. Feature Engineering
   -> AgeGroup
   -> CapitalNet
   -> HasCapitalGain
   -> WorkHoursCategory
   -> EducationLevel

7. Feature Analysis
   -> Study relationships between engineered features
      and income.

8. Model Preparation
   -> Encode categorical variables.
   -> Scale numerical features when required.
   -> Split data into training and testing sets.

9. Model Building
   -> Logistic Regression
   -> Decision Tree
   -> Random Forest

10. Model Evaluation
   -> Accuracy
   -> Precision
   -> Recall
   -> F1-score
   -> Confusion Matrix

```
